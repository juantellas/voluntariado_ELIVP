# 2. Inicio de proceso de ETL (Filtrado)

```python
df = pd.read_excel(r"C:\Users\taran\OneDrive\Documents\uni\voluntariado\Postulación+ELIVP+2026_May+4,+2026_13.59_filtradotrue.xlsx")
```

```python
df = df.drop(columns=["StartDate", "EndDate", "Status", "IPAddress", "Progress", "Duration (in seconds)", 
                      "RecordedDate", "ResponseId","RecipientLastName", "RecipientFirstName", "DistributionChannel", "UserLanguage","Q_RecaptchaScore","Q_RecaptchaStatus",
                      "Q_RecaptchaError","RecipientEmail","ExternalReference"])
```

```python
df.head()
```

```python
import pandas as pd
%matplotlib inline
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(20, 9))
sns.heatmap(df.isna(), cmap='Purples', cbar=False, yticklabels=False)

plt.title(' Mapa de calor de valores faltantes')
plt.xlabel('Variables')
plt.tight_layout()
plt.show()
```

```python
df["Q13"] = df["Q13"].fillna("Ninguna")
df["Q28"] = df["Q28"].fillna("N/A")
df["Q29"] = df["Q29"].fillna("N/A")
df["Q30"] = df["Q30"].fillna("Ninguna")
df["Q35"] = df["Q35"].fillna("N/A")
df["Q16"] = df["Q16"].fillna("N/A")
df["Q17"] = df["Q17"].fillna("N/A")
df["Q38"] = df["Q38"].fillna("N/A")
df["Q36"] = df["Q36"].fillna("N/A")
```

```python
cols_barrio = ["Q20", "Q21", "Q22", "Q23", "Q24"]

df["Q20"] = df[cols_barrio].stack().groupby(level=0).first()
```

```python
df = df.drop(columns=["Q10" ,"Q21", "Q22", "Q23", "Q24","Q41","LocationLatitude","LocationLongitude" ])
```

```python
df = df.iloc[1:]
df.head()
```

```python
import pandas as pd
%matplotlib inline
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(20, 9))
sns.heatmap(df.isna(), cmap='Purples', cbar=False, yticklabels=False)

plt.title(' Mapa de calor de valores faltantes')
plt.xlabel('Variables')
plt.tight_layout()
plt.show()
```

```python
# Tabla de resumen de valores faltantes
missing_summary = pd.DataFrame({
    'Variable': df.columns,
    'Faltantes (#)': df.isna().sum().values,
    'Faltantes (%)': (df.isna().mean().values * 100).round(2)
})

# Filtrar solo variables con al menos un dato faltante
missing_summary = missing_summary[missing_summary['Faltantes (#)'] > 0]

# Ordenar por porcentaje descendente
missing_summary = missing_summary.sort_values(by='Faltantes (%)', ascending=False).reset_index(drop=True)

# Mostrar tabla
print(missing_summary)
```

```python
cols = ["Q16", "Q17", "Q38", "Q20", "Q36"]

df[df[cols].isna().any(axis=1)]
```

```python
df["Q20"] = df["Q20"].fillna("N/A")
```
